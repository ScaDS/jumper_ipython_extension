# AI-Powered Optimization Review

Quick examples of `%perfmonitor_ai_review`. Requires the optional `[ai]` extras:

```
pip install jumper-extension[ai]
```

````python
%load_ext jumper_extension
````

````python
%perfmonitor_start 1
````

````python
import time

# Some workload to analyze
values = [i ** 2 for i in range(1_000_000)]
time.sleep(1)
sum(values[:10_000])
````

Analyze the last cell and get ranked optimization suggestions with a run id:

````python
%perfmonitor_ai_review
````

Analyze a specific cell range at a different level:

Apply a suggestion from a previous run (replace `RUN_ID` with the id printed above):

````python
%perfmonitor_ai_review --resume a577d45d --select 3
````

````python
import time
import numpy as np

values = np.square(np.arange(1, 1_000_000 + 1))
time.sleep(1)
sum(values[:10_000])
````

Or apply a refined version of a suggestion with a custom instruction:

````python
%perfmonitor_ai_review --resume RUN_ID --select 1 --refine "use multiprocessing instead of joblib"
````

````python
%perfmonitor_stop
````
