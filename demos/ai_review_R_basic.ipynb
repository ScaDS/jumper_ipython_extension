# AI Reviewer — R basic

Setup, then one imperfect R computation for `%perfmonitor_ai_review --benchmark`.

```r
set.seed(42)
n <- 700
d <- 25
X <- matrix(rnorm(n * d), nrow = n, ncol = d)
```

```r
pairwise_dist <- function(X) {
  n <- nrow(X)
  D <- matrix(0, n, n)
  for (i in 1:n) {
    for (j in 1:n) {
      s <- 0
      for (k in 1:ncol(X)) {
        s <- s + (X[i, k] - X[j, k])^2
      }
      D[i, j] <- sqrt(s)
    }
  }
  D
}

D <- pairwise_dist(X)
neighbors <- t(apply(D, 1, order))[, 2:6]
```
