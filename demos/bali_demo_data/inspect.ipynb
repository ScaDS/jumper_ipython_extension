```python
import os
import shutil
from pathlib import Path

def create_mock_data(input_pid):
    current_pid = os.getpid()
    mock_pid = f"{current_pid}"
    # Create mock directories
    for base_dir in ["perfdata_results", "bali_results"]:
        src_path = Path(base_dir) / str(input_pid)
        dst_path = Path(base_dir) / mock_pid
        
        if src_path.exists():
            if dst_path.exists():
                shutil.rmtree(dst_path)
            shutil.copytree(src_path, dst_path)
            print(f"Copied {src_path} -> {dst_path}")
        else:
            print(f"Source {src_path} not found")
    
    return mock_pid
    
def cleanup_mock_data():
    target_pid = str(os.getpid())
    removed_any = False
    for base_dir in ["perfdata_results", "bali_results"]:
        target_path = Path(base_dir) / target_pid
        if target_path.exists():
            shutil.rmtree(target_path)
            print(f"Removed {target_path}")
            removed_any = True
    if not removed_any:
        print(f"No mock data found for PID {target_pid}")
```
